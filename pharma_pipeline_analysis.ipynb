```python
submissions = pd.read_csv(
    os.path.join(FDA_PATH, 'Submissions.txt'),
    sep='\t',
    encoding='latin-1',
    on_bad_lines='skip'
)

print(f"Submissions shape: {submissions.shape}")
print(f"Columns: {submissions.columns.tolist()}")
print(submissions.head(3))
```

```python
# ── STEP 1: FILTER SUBMISSIONS TO APPROVALS ONLY ────────────
approved = submissions[submissions['SubmissionStatus'] == 'AP'].copy()
print(f"✓ Approved submissions: {approved.shape[0]:,} rows")

# ── STEP 2: PARSE THE DATE ───────────────────────────────────
approved['ActionDate'] = pd.to_datetime(
    approved['SubmissionStatusDate'], errors='coerce'
)
approved['ApprovalYear'] = approved['ActionDate'].dt.year
approved['ApprovalMonth'] = approved['ActionDate'].dt.month

# ── STEP 3: THREE-WAY MERGE ──────────────────────────────────
# Products → Applications → Submissions
df = products.merge(applications, on='ApplNo', how='left')
df = df.merge(
    approved[['ApplNo', 'SubmissionType', 'ActionDate', 'ApprovalYear', 'ApprovalMonth']],
    on='ApplNo',
    how='left'
)

print(f"✓ After merge: {df.shape[0]:,} rows")

# ── STEP 4: DROP NULLS IN KEY COLUMNS ────────────────────────
df = df.dropna(subset=['ActionDate', 'SponsorName', 'ActiveIngredient'])
print(f"✓ After dropping nulls: {df.shape[0]:,} rows")

# ── STEP 5: FILTER TO 2000 ONWARDS ───────────────────────────
df = df[df['ApprovalYear'] >= 2000]
print(f"✓ After filtering to 2000+: {df.shape[0]:,} rows")

# ── STEP 6: RENAME SubmissionType → ApplType FOR CONSISTENCY ─
# ORIG = Original application (new drug), SUPPL = Supplement
# We'll use ApplType from Applications for NDA/BLA/ANDA labels
print(f"\nApplType values: {df['ApplType'].value_counts().to_dict()}")
print(f"SubmissionType values: {df['SubmissionType'].value_counts().to_dict()}")

# ── STEP 7: CLEAN COMPANY NAMES ──────────────────────────────
df['SponsorName'] = df['SponsorName'].str.upper().str.strip()

name_map = {
    'PFIZER INC': 'PFIZER',
    'PFIZER PHARMACEUTICALS': 'PFIZER',
    'PFIZER PHARMACEUTICALS INC': 'PFIZER',
    'NOVARTIS PHARMACEUTICALS': 'NOVARTIS',
    'NOVARTIS PHARMA AG': 'NOVARTIS',
    'NOVARTIS PHARMACEUTICALS CORP': 'NOVARTIS',
    'JOHNSON AND JOHNSON': 'J&J',
    'JOHNSON & JOHNSON': 'J&J',
    'JANSSEN PHARMACEUTICA': 'J&J',
    'JANSSEN PHARMACEUTICALS INC': 'J&J',
    'ASTRAZENECA PHARMACEUTICALS': 'ASTRAZENECA',
    'ASTRAZENECA LP': 'ASTRAZENECA',
    'ELI LILLY AND COMPANY': 'ELI LILLY',
    'ELI LILLY & CO': 'ELI LILLY',
    'BRISTOL-MYERS SQUIBB': 'BMS',
    'BRISTOL MYERS SQUIBB': 'BMS',
    'MERCK SHARP & DOHME': 'MERCK',
    'MERCK & CO INC': 'MERCK',
    'ROCHE LABORATORIES': 'ROCHE',
    'F. HOFFMANN-LA ROCHE': 'ROCHE',
    'SANOFI-AVENTIS': 'SANOFI',
    'SANOFI AVENTIS': 'SANOFI',
}
df['SponsorName'] = df['SponsorName'].replace(name_map)

# ── STEP 8: SUMMARY ──────────────────────────────────────────
print(f"\n✓ Clean dataset ready")
print(f"  Year range:       {int(df['ApprovalYear'].min())} – {int(df['ApprovalYear'].max())}")
print(f"  Unique companies: {df['SponsorName'].nunique():,}")
print(f"  Unique molecules: {df['ActiveIngredient'].nunique():,}")
print(f"  Total rows:       {df.shape[0]:,}")
```

```python
# ── DEDUPLICATE: KEEP ONLY ORIGINAL APPROVALS (ORIG) ────────
# Each drug has 1 ORIG (first approval) + many SUPPLs (label updates, new dosages etc)
# For market analysis we want to count each drug ONCE — its first approval

df_orig = df[df['SubmissionType'] == 'ORIG'].copy()

# If somehow a drug still has duplicates, keep the earliest approval
df_orig = df_orig.sort_values('ActionDate')
df_orig = df_orig.drop_duplicates(subset=['ApplNo'], keep='first')

print(f"✓ Original approvals only (unique drugs): {df_orig.shape[0]:,} rows")
print(f"  NDA (brand):  {(df_orig['ApplType']=='NDA').sum():,}")
print(f"  ANDA (generic): {(df_orig['ApplType']=='ANDA').sum():,}")
print(f"  BLA (biologic): {(df_orig['ApplType']=='BLA').sum():,}")
print(f"\n  Year range: {int(df_orig['ApprovalYear'].min())} – {int(df_orig['ApprovalYear'].max())}")
```

```python
yearly = (df_orig.groupby(['ApprovalYear', 'ApplType'])
    .size()
    .reset_index(name='approvals')
)

yearly.to_csv(os.path.join(OUTPUT_PATH, 'yearly_approvals.csv'), index=False)
print(f"✓ yearly_approvals.csv — {len(yearly)} rows")
yearly[yearly['ApprovalYear'] >= 2020]
```

```python
company = (df_orig.groupby(['SponsorName', 'ApplType'])
    .agg(
        total_approvals=('ApplNo', 'count'),
        unique_molecules=('ActiveIngredient', 'nunique'),
        latest_approval=('ApprovalYear', 'max')
    )
    .reset_index()
    .sort_values('total_approvals', ascending=False)
)

company.to_csv(os.path.join(OUTPUT_PATH, 'company_performance.csv'), index=False)
print(f"✓ company_performance.csv — {len(company)} rows")
company.head(10)
```

```python
df_orig['FormClean'] = df_orig['Form'].str.upper().str.extract(
    r'(TABLET|CAPSULE|INJECTION|SOLUTION|CREAM|PATCH|INHALER|POWDER|GEL|SPRAY)'
)

form_trend = (df_orig.groupby(['ApprovalYear', 'FormClean'])
    .size()
    .reset_index(name='count')
    .dropna()
)

form_trend.to_csv(os.path.join(OUTPUT_PATH, 'form_trends.csv'), index=False)
print(f"✓ form_trends.csv — {len(form_trend)} rows")
form_trend.head(10)
```

```python
velocity = (df_orig.groupby(['SponsorName', 'ApprovalYear'])
    .size()
    .reset_index(name='approvals_that_year')
)

velocity = velocity.sort_values(['SponsorName', 'ApprovalYear'])
velocity['rolling_3yr_avg'] = (
    velocity.groupby('SponsorName')['approvals_that_year']
    .transform(lambda x: x.rolling(3, min_periods=1).mean())
    .round(2)
)

velocity.to_csv(os.path.join(OUTPUT_PATH, 'approval_velocity.csv'), index=False)
print(f"✓ approval_velocity.csv — {len(velocity)} rows")

# Preview top companies
top_companies = company.head(5)['SponsorName'].tolist()
velocity[velocity['SponsorName'].isin(top_companies)].tail(15)
```

```python
# ── FASTER VERSION: BUILD VELOCITY FROM df_orig DIRECTLY ────

# Get top 50 companies first (fast operation)
top_50_companies = (df_orig.groupby('SponsorName')['ApplNo']
    .count()
    .sort_values(ascending=False)
    .head(50)
    .index.tolist()
)

# Filter df_orig to top 50 companies only, then build velocity
velocity_filtered = (df_orig[df_orig['SponsorName'].isin(top_50_companies)]
    .groupby(['SponsorName', 'ApprovalYear'])
    .size()
    .reset_index(name='approvals_that_year')
)

velocity_filtered = velocity_filtered.sort_values(['SponsorName', 'ApprovalYear'])
velocity_filtered['rolling_3yr_avg'] = (
    velocity_filtered.groupby('SponsorName')['approvals_that_year']
    .transform(lambda x: x.rolling(3, min_periods=1).mean())
    .round(2)
)

velocity_filtered.to_csv(
    os.path.join(OUTPUT_PATH, 'approval_velocity.csv'),
    index=False
)

print(f"✓ Velocity file saved")
print(f"  Rows: {len(velocity_filtered):,}")
print(f"  Companies: {velocity_filtered['SponsorName'].nunique()}")
print(f"\nTop 10 companies:")
print(top_50_companies[:10])
```

```python
# ── ORANGE BOOK PATENT CLIFF ANALYSIS ───────────────────────

# Parse approval date
orange['Approval_Date'] = pd.to_datetime(orange['Approval_Date'], errors='coerce')
orange['ApprovalYear'] = orange['Approval_Date'].dt.year

# Filter to only NDA/BLA (brand drugs) — generics don't have patent cliffs
brand_orange = orange[orange['Appl_Type'].isin(['N', 'B'])].copy()

# We'll use Approval_Date as a proxy for patent expiry
# Patents last 20 years from filing; market exclusivity ~5 years from approval
# Simple proxy: approval year + 5 years exclusivity window
brand_orange['EstimatedExpiryYear'] = brand_orange['ApprovalYear'] + 5

# Filter to drugs whose exclusivity expires 2024-2030
patent_cliff = (brand_orange[
    brand_orange['EstimatedExpiryYear'].between(2024, 2030)
]
.groupby('EstimatedExpiryYear')
.agg(
    drugs_losing_exclusivity=('Ingredient', 'nunique'),
    example_drugs=('Trade_Name', lambda x: ', '.join(x.dropna().unique()[:3]))
)
.reset_index()
.rename(columns={'EstimatedExpiryYear': 'ExpiryYear'})
)

patent_cliff.to_csv(os.path.join(OUTPUT_PATH, 'patent_cliff.csv'), index=False)

print(f"✓ patent_cliff.csv saved")
print(f"\nPatent cliff breakdown 2024–2030:")
print(patent_cliff.to_string(index=False))
print(f"\nTotal drugs losing exclusivity: {patent_cliff['drugs_losing_exclusivity'].sum()}")
```
